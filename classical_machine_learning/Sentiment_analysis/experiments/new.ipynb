## Text Preprocessing: Removing Stopwords from Lemmatized Reviews

```python
import pandas as pd
import re
from nltk.corpus import stopwords

import spacy

# Load SpaCy tokenizer only (fast)
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])

# Use SpaCy’s built-in stopwords set
stop_words = nlp.Defaults.stop_words

# Function to remove stopwords (fast, plain text)
def remove_stopwords(text):
    return " ".join([word for word in str(text).split() if word.lower() not in stop_words])


combined_data=pd.read_csv("../data/combined_lemmatized.csv",encoding='latin1')


# Remove stopwords from the 'review' column
combined_data['review'] = combined_data['review'].apply(remove_stopwords)

# Save new CSV
combined_data.to_csv("imdb_no_stopwords.csv", index=False)
```
