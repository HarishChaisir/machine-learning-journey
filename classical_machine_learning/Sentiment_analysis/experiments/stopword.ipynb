```python
import numpy as np
import pandas as pd

combined_data=pd.read_csv("../data/combined_lemmatized.csv",encoding='latin1')
```

```python

import spacy

# Load SpaCy with only the tokenizer (disable parser, NER for speed)
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])

def remove_stopwords_batch(texts, batch_size=100):
    """
    Removes stopwords from a list of texts.
    Returns a list of strings with stopwords removed.
    """
    processed_texts = []
    for doc in nlp.pipe(texts, batch_size=batch_size):
        tokens = [token.text for token in doc if not token.is_stop]
        processed_texts.append(" ".join(tokens))
    return processed_texts
```

```python
combined_data['review'] = remove_stopwords_batch(combined_data['review'].astype(str).tolist(), batch_size=100)
```

```python
combined_data.to_csv("output_no_stopwords.csv", index=False)
```
