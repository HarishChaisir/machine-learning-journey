```python
# spam_env\Scripts\Activate
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report
```

```python
spam_classifier_data = pd.read_csv('D:/Text_classifier/Spam classifier/data/spam.csv',encoding='latin1')
```

```python
spam_classifier_data.head()
```

```python
spam_classifier_data=spam_classifier_data.drop(spam_classifier_data.columns[2:5],axis=1)
```

```python
spam_classifier_data.head()
```

```python
spam_classifier_data=spam_classifier_data.dropna(axis=0)
```

```python
spam_classifier_data.tail()
```

```python
spam_classifier_data=spam_classifier_data.drop_duplicates()
```

```python
spam_classifier_data.tail()
```

```python
spam_classifier_data['v1'].value_counts()
```

```python
spam_classifier_data.shape
```

```python
spam_classifier_data['v2']=spam_classifier_data['v2'].str.strip()
spam_classifier_data['v2'] = spam_classifier_data['v2'].str.replace(r'\s+', ' ', regex=True)
spam_classifier_data['v2']=spam_classifier_data['v2'].str.lower()
spam_classifier_data['v2'] = spam_classifier_data['v2'].str.replace(r'[^a-zA-Z0-9.,!? ]', '', regex=True)


```

```python
spam_classifier_data.head(100)
```

```python
spam_classifier_data.tail()
```

```python
vectorizer=TfidfVectorizer()
X=vectorizer.fit_transform(spam_classifier_data['v2'])
Y=spam_classifier_data['v1']
```

```python
print(X)
```

```python
X_train, X_test, Y_train, Y_test= train_test_split(X,Y,test_size=0.1, stratify=Y, random_state=1)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
model=MultinomialNB()
```

```python
model.fit(X_train,Y_train)
```

```python
X_train_prediction=model.predict(X_train)
training_data_accuracy=accuracy_score(X_train_prediction,Y_train)

print("Accuracy on training data:",training_data_accuracy)
```

```python
X_test_prediction=model.predict(X_test)
test_data_accuracy=accuracy_score(X_test_prediction,Y_test)

print("Accuracy on testing data:",test_data_accuracy)
```

```python

print(classification_report(Y_test, X_test_prediction))
```
